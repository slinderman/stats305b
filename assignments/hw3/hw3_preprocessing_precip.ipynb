# Preprocessing Climate Data

Download the preciptation data from NOAA (https://www.ncei.noaa.gov/products/climate-data-records/precipitation-gpcp-monthly). 

**Climate Data Record Specifications**
- Global coverage 
- 2.5 × 2.5 degree  product resolution 
- 1979–present 
- Global coverage 
- Preliminary data updated monthly 
- Final data updated every 3 months

Example file name
```
https://www.ncei.noaa.gov/data/global-precipitation-climatology-project-gpcp-monthly/access/1979/gpcp_v02r03_monthly_d197901_c20170616.nc
```

````python
for i, fname in enumerate(fnames):
    year = 2017 + (i // 12)
    url = f"https://www.ncei.noaa.gov/data/global-precipitation-climatology-project-gpcp-monthly/access/{str(year)}/{fname}"
    if os.path.exists(f"./data/{fname}"):
        continue

    try:
        print(f"Downloading {url}")
        urllib.request.urlretrieve(url, f"./data/{fname}")
    except Exception as e:
        print(f"Error downloading file: {e}")
````

## Load all the netCDF4 files and extract the precip data

````python
years = []
months = []
precips = []

for year in range(1979, 2025):
    for month in range(1, 13):
        matches = glob.glob(f"./data/gpcp_v02r03_monthly_d{str(year)}{str(month).zfill(2)}*.nc")
        
        if len(matches) == 0:
            break
        
        dataset = netCDF4.Dataset(matches[0])
        years.append(year)
        months.append(month)
        precips.append(np.array(dataset.variables["precip"][:]))
    
precips = np.concatenate(precips, axis=0)
years = np.array(years)
months = np.array(months)
num_timesteps = len(years)
````

### Extract the locations

````python
dataset.variables['precip']
````

````python
latitude = np.array(dataset.variables['latitude'][:]) 
longitude = np.array(dataset.variables['longitude'][:])
lons, lats = np.meshgrid(longitude, latitude)
num_locs = len(longitude) * len(latitude)
````

### Package into dataframe

````python
# Flatten the precipitation data so each location is a row
df = pd.DataFrame(dict(
    latitude=np.tile(lats.ravel(), num_timesteps),
    longitude=np.tile(lons.ravel(), num_timesteps),
    year=np.repeat(years, num_locs),
    month=np.repeat(months, num_locs),
    day=np.ones(num_timesteps * num_locs, dtype=int),
    precip=np.ravel(precips)
    ))

df["date"] = pd.to_datetime(df[["year", "month", "day"]])

month_names = {
    1: "January",
    2: "February",
    3: "March",
    4: "April",
    5: "May",
    6: "June",
    7: "July",
    8: "August",
    9: "September",
    10: "October",
    11: "November",
    12: "December"
}

df
````

````python
import plotly.express as px

def plot_precipitation(year, month, interactive=True):
    fig = px.density_map(df[(df.year==year) & (df.month==month)], 
                         lat='latitude', 
                         lon='longitude', 
                         z='precip', 
                         radius=8,
                         center=dict(lat=0, lon=0), 
                         zoom=0.1,
                         range_color=(0, 25),
                         opacity=0.8,
                         color_continuous_scale="Purp",
                         width=800,
                         height=600,
                         map_style="carto-positron",
                         labels=dict(precip="precipitation (mm/day)"),
                         )
    
    fig.update_layout(
        coloraxis_colorbar_title_side="right",
        title={
            'text': f"Precipitation in {month_names[month]}, {year}",
            'x': 0.5,
            'xanchor': 'center'
        },
        hovermode=False)
    
    if interactive:
        return fig
    else:
        return Image(fig.to_image(scale=1.0))
    
fig = plot_precipitation(2016, 7, interactive=True)
fig.show()
````

````python
num_timesteps = masked_temps.shape[0]
lons, lats = np.meshgrid(temp_dataset.variables["longitude"][:], 
                         temp_dataset.variables["latitude"][:])
num_locs = lons.size

````

````python
# Flatten the temperature data so each location is a row
offsets = pd.Series([pd.Timedelta(days=i) for i in temp_dataset.variables["time"][:].data])

kelvin_to_celcius = lambda x: x - 273.15

temp_df = pd.DataFrame(dict(
    date=pd.Timestamp(year=1978, month=1, day=1) + np.repeat(offsets, num_locs),
    latitude=np.tile(lats.ravel(), num_timesteps),
    longitude=np.tile(lons.ravel(), num_timesteps),
    temperature=kelvin_to_celcius(np.ravel(temp_dataset.variables["tcdr_MSU_AMSUA_ATMS_TLT"][:], order='C'))
    ))

# temp_df[np.isfinite(temp_df.temp)]
# temp_df.dropna(axis=0, how='any', inplace=True)
temp_df.reset_index(drop=True, inplace=True)
temp_df
````

````python
import plotly.express as px
import plotly.figure_factory as ff

def plot_temperature(df, year, month, interactive=True):
    diff = (df.date - pd.Timestamp(year=year, month=month, day=16)).abs()
    nearest_date = df.loc[diff.idxmin()].date

    fig = px.density_map(df[df.date == nearest_date], 
                         lat='latitude', 
                         lon='longitude', 
                         z='temperature', 
                         radius=8,
                         center=dict(lat=0, lon=0), 
                         zoom=0.3,
                         range_color=(-30, 10),
                         opacity=0.8,
                         color_continuous_scale="thermal",
                         width=800,
                         height=600,
                         map_style="carto-positron",
                         labels=dict(precip="precipitation (mm/day)"),
                         )
    
    fig.update_layout(
        coloraxis_colorbar_title_side="right",
        title={
            'text': f"temperature (C) in {month_names[month]}, {year}",
            'x': 0.5,
            'xanchor': 'center'
        })
    fig.show()

    if interactive:
        return fig
    else:
        return Image(fig.to_image(scale=1.0))
    
# fig, nearest_date = 
fig = plot_temperature(temp_df, 2016, 1, interactive=True)
# fig.show()
````
