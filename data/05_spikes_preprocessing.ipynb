# Preprocess the data from Jonathan Pillow's tutorial into a CSV
See https://github.com/pillowlab/GLMspiketraintutorial for the tutorial.


```python
# Download the data and unzip it
!wget https://pillowlab.princeton.edu/data/data_RGCs.zip
!unzip -n data_RGCs.zip 
```

```python
from scipy.io import loadmat
import numpy as np
import pandas as pd

# Preprocess the data into features and responses
stim = loadmat('data_RGCs/Stim.mat')['Stim']
stimtimes = loadmat('data_RGCs/stimtimes.mat')['stimtimes'][:, 0]
dt = stimtimes[1] - stimtimes[0]
spikecounts = np.zeros((len(stim), 4), dtype=int)
for i in range(4):
    sptimes = loadmat('data_RGCs/SpTimes.mat')['SpTimes'][0, i][:, 0]
    spikecounts[(sptimes // dt).astype(int), i] = 1

# Save as a pandas dataframe, but only keep the first 10000 frames to save memory
T = 10000
df = pd.DataFrame(dict(t=stimtimes[:T], 
                       stim=stim[:T, 0], 
                       y0=spikecounts[:T, 0], 
                       y1=spikecounts[:T, 1], 
                       y2=spikecounts[:T, 2], 
                       y3=spikecounts[:T, 3]))
df.to_csv('05_spikes.csv', index=False)
```
